```python
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from transformers import pipeline
import re
```

```python
def get_video_id(url):
    match = re.search(r"(?:v=|youtu\.be/)([A-Za-z0-9_-]{11})", url)
    if not match:
        raise ValueError("Couldn't extract video ID. Paste a valid YouTube URL.")
    return match.group(1)

def get_transcript(url):
    video_id = get_video_id(url)
    transcript_list = YouTubeTranscriptApi.get_transcript(video_id)
    full_text = " ".join([entry['text'] for entry in transcript_list])
    return full_text
```

```python
def get_transcript(url):
    video_id = get_video_id(url)
    ytt_api = YouTubeTranscriptApi()
    transcript_list = ytt_api.fetch(video_id)
    full_text = " ".join([entry.text for entry in transcript_list])
    return full_text
```

```python
url = "https://www.youtube.com/watch?v=pQbvrDl3kcY&list=RDpQbvrDl3kcY&start_radio=1"
transcript = get_transcript(url)
print(transcript[:1000])
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_text(transcript)
print(f"Number of chunks: {len(chunks)}")
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectorstore = Chroma.from_texts(chunks, embedding_model)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")
```

```python
def generate_text(prompt, max_length=200):
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512)
    outputs = model.generate(**inputs, max_length=max_length)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
print(generate_text("Summarize: The cat sat on the mat and looked out the window all day."))
```

```python
def ask_video(question, k=3):
    docs = vectorstore.similarity_search(question, k=k)
    context = "\n\n".join([doc.page_content for doc in docs])

    prompt = f"""Answer the question using only the video transcript context below.

Context:
{context}

Question: {question}
Answer:"""

    return generate_text(prompt, max_length=200)
```

```python
def summarize_video(text, max_chunk_chars=3000):
    pieces = [text[i:i+max_chunk_chars] for i in range(0, len(text), max_chunk_chars)]
    summaries = []
    for piece in pieces:
        prompt = f"Summarize this transcript excerpt:\n\n{piece}"
        summaries.append(generate_text(prompt, max_length=150))

    combined = " ".join(summaries)
    if len(pieces) > 1:
        final_prompt = f"Combine these summaries into one coherent summary:\n\n{combined}"
        return generate_text(final_prompt, max_length=200)
    return combined
```

```python
url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ" # Updated to a video that is available and has transcripts

# Step A: Get transcript
transcript = get_transcript(url)

# Step B: Build vector store
chunks = splitter.split_text(transcript)
vectorstore = Chroma.from_texts(chunks, embedding_model)

# Step C: Summarize
print("SUMMARY:")
print(summarize_video(transcript))

# Step D: Ask questions
print("\nQ&A:")
print(ask_video("What is the main topic of this video?"))
print(ask_video("What key points are mentioned?"))
```

```python
from gtts import gTTS
from IPython.display import Audio, display

url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ" # Updated to a video that is available and has transcripts

# Step A: Get transcript
transcript = get_transcript(url)

# Step B: Build vector store
chunks = splitter.split_text(transcript)
vectorstore = Chroma.from_texts(chunks, embedding_model)

# Step C: Summarize (text)
summary_text = summarize_video(transcript)
print("SUMMARY (text):")
print(summary_text)

# Step D: Convert summary to speech
tts = gTTS(text=summary_text, lang='en')
tts.save("summary.mp3")

# Step E: Play the audio directly in Colab
display(Audio("summary.mp3", autoplay=False))
```

```python
def speak(text, filename="answer.mp3"):
    tts = gTTS(text=text, lang='en')
    tts.save(filename)
    display(Audio(filename, autoplay=False))

# Example usage:
answer = ask_video("What is the main topic of this video?")
print(answer)
speak(answer)
```
